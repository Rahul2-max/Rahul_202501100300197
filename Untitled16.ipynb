```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score
from sklearn.neighbors import NearestNeighbors

import warnings
warnings.filterwarnings("ignore")
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("data.csv", encoding="ISO-8859-1")
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

print(df.shape)
df.head()
```

```python
print("Missing values:\n", df.isnull().sum(), "\n")

df = df.dropna(subset=["CustomerID"])                     # RFM needs a customer
df = df[~df["InvoiceNo"].astype(str).str.startswith("C")] # remove cancellations
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]     # remove returns / bad prices
df = df.drop_duplicates()

df["CustomerID"] = df["CustomerID"].astype(int)
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]

print("Clean shape:", df.shape)
print("Unique customers:", df["CustomerID"].nunique())
```

```python
# Reference date = one day after the last transaction
snapshot_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

rfm = df.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x: (snapshot_date - x.max()).days),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("TotalPrice", "sum"),
)

print(rfm.describe())
rfm.head()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 3.5))
for ax, col in zip(axes, rfm.columns):
    sns.histplot(rfm[col], bins=40, ax=ax)
    ax.set_title(f"{col} (raw)")
plt.tight_layout()
plt.show()

rfm_log = np.log1p(rfm)
scaler = StandardScaler()
X = scaler.fit_transform(rfm_log)
```

```python
k_range = range(2, 11)
inertias, sil_scores = [], []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X, labels))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(k_range, inertias, "o-")
axes[0].set(title="Elbow Method", xlabel="K", ylabel="Inertia")
axes[1].plot(k_range, sil_scores, "o-", color="green")
axes[1].set(title="Silhouette Score", xlabel="K", ylabel="Score")
plt.tight_layout()
plt.show()

print(dict(zip(k_range, np.round(sil_scores, 3))))
```

```python
K = 4   # change this if your elbow/silhouette plots clearly suggest another value

kmeans = KMeans(n_clusters=K, random_state=42, n_init=10)
rfm["KMeans_Cluster"] = kmeans.fit_predict(X)

print("K-Means silhouette:", round(silhouette_score(X, rfm["KMeans_Cluster"]), 3))
print(rfm["KMeans_Cluster"].value_counts().sort_index())
```

```python
min_samples = 10

nn = NearestNeighbors(n_neighbors=min_samples).fit(X)
distances, _ = nn.kneighbors(X)
k_dist = np.sort(distances[:, -1])

plt.figure(figsize=(7, 4))
plt.plot(k_dist)
plt.title("k-distance plot (look for the elbow)")
plt.xlabel("Points sorted by distance")
plt.ylabel(f"Distance to {min_samples}th neighbour")
plt.show()
```

```python
eps = 0.5   # adjust to the elbow you see in the plot above

dbscan = DBSCAN(eps=eps, min_samples=min_samples)
rfm["DBSCAN_Cluster"] = dbscan.fit_predict(X)

n_clusters = len(set(rfm["DBSCAN_Cluster"])) - (1 if -1 in rfm["DBSCAN_Cluster"].values else 0)
n_noise = (rfm["DBSCAN_Cluster"] == -1).sum()
print("Clusters found:", n_clusters, "| Noise points:", n_noise)

mask = rfm["DBSCAN_Cluster"] != -1
if n_clusters > 1:
    print("DBSCAN silhouette (excluding noise):",
          round(silhouette_score(X[mask], rfm.loc[mask, "DBSCAN_Cluster"]), 3))
print(rfm["DBSCAN_Cluster"].value_counts().sort_index())
```

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X)

print("Explained variance ratio:", np.round(pca.explained_variance_ratio_, 3))
print("Total variance explained:", round(pca.explained_variance_ratio_.sum(), 3))

rfm["PCA1"], rfm["PCA2"] = X_pca[:, 0], X_pca[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=rfm, x="PCA1", y="PCA2", hue="KMeans_Cluster",
                palette="tab10", s=20, ax=axes[0])
axes[0].set_title("PCA: K-Means clusters")
sns.scatterplot(data=rfm, x="PCA1", y="PCA2", hue="DBSCAN_Cluster",
                palette="tab10", s=20, ax=axes[1])
axes[1].set_title("PCA: DBSCAN clusters (-1 = noise)")
plt.tight_layout()
plt.show()
```

```python
tsne = TSNE(n_components=2, perplexity=30, init="pca",
            learning_rate="auto", random_state=42)
X_tsne = tsne.fit_transform(X)

rfm["TSNE1"], rfm["TSNE2"] = X_tsne[:, 0], X_tsne[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=rfm, x="TSNE1", y="TSNE2", hue="KMeans_Cluster",
                palette="tab10", s=20, ax=axes[0])
axes[0].set_title("t-SNE: K-Means clusters")
sns.scatterplot(data=rfm, x="TSNE1", y="TSNE2", hue="DBSCAN_Cluster",
                palette="tab10", s=20, ax=axes[1])
axes[1].set_title("t-SNE: DBSCAN clusters (-1 = noise)")
plt.tight_layout()
plt.show()
```

```python
profile = rfm.groupby("KMeans_Cluster").agg(
    Customers=("Recency", "size"),
    Avg_Recency=("Recency", "mean"),
    Avg_Frequency=("Frequency", "mean"),
    Avg_Monetary=("Monetary", "mean"),
    Total_Revenue=("Monetary", "sum"),
).round(1)

profile["Revenue_%"] = (profile["Total_Revenue"] / profile["Total_Revenue"].sum() * 100).round(1)
profile["Customer_%"] = (profile["Customers"] / profile["Customers"].sum() * 100).round(1)
profile
```

```python
# Visual comparison of cluster averages
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, ["Recency", "Frequency", "Monetary"]):
    sns.barplot(data=rfm.groupby("KMeans_Cluster")[col].mean().reset_index(),
                x="KMeans_Cluster", y=col, palette="tab10", ax=ax)
    ax.set_title(f"Average {col}")
plt.tight_layout()
plt.show()
```

```python
med = rfm[["Recency", "Frequency", "Monetary"]].median()

personas = {
    (True,  True,  True):  ("Champions",
        "Reward with VIP perks, early access to new products, referral bonuses."),
    (True,  True,  False): ("Loyal Regulars",
        "Upsell and cross-sell, bundle offers, loyalty points."),
    (True,  False, True):  ("Big-Basket Newcomers",
        "Encourage a second purchase with follow-up offers and a welcome series."),
    (True,  False, False): ("New / Low-Spend Customers",
        "Onboarding emails, small-discount incentives, product recommendations."),
    (False, True,  True):  ("At-Risk High-Value",
        "Urgent win-back: personalised offers, a call or email from account management."),
    (False, True,  False): ("At-Risk Regulars",
        "Reminder emails, 'we miss you' discounts, highlight new arrivals."),
    (False, False, True):  ("Lapsed Big Spenders",
        "Targeted reactivation with premium or bulk-purchase deals."),
    (False, False, False): ("Lost / Hibernating",
        "Low-cost last-chance campaign; otherwise suppress to save marketing spend."),
}

means = rfm.groupby("KMeans_Cluster")[["Recency", "Frequency", "Monetary"]].mean()

rows = []
for c, r in means.iterrows():
    key = (r["Recency"] < med["Recency"],     # recent = good
           r["Frequency"] > med["Frequency"],
           r["Monetary"] > med["Monetary"])
    name, action = personas[key]
    rows.append({"Cluster": c, "Persona": name,
                 "Avg_R": round(r["Recency"], 1),
                 "Avg_F": round(r["Frequency"], 1),
                 "Avg_M": round(r["Monetary"], 1),
                 "Marketing Action": action})

persona_df = pd.DataFrame(rows)
persona_df
```
